```python
import pandas as pd
from sqlalchemy import create_engine

# Load CSV
df = pd.read_csv('flights_data.csv')

# Connect to MySQL
engine = create_engine("mysql+pymysql://root:1234@localhost/flights_db")

# Push DataFrame to MySQL (automatic column mapping)
df.to_sql('flights', con=engine, if_exists='append', index=False)
```
